# ULSD CIF Window Report

Rebuilds **NWE/Med ULSD Window** from eWindow with **day-driven discovery** (not hard-coded hubs).

| Section | Contents |
|---------|----------|
| **CIF NWE** | MOPL (`AMOPL00`), hub volume + Sankey, all NWE PVO basis charts that day |
| **CIF Med** | MOPL (`AMOPN00`), hub volume + Sankey, all Med PVO basis charts |
| **FOB ARA** | Cargo volume + Sankey, FOB Basis ARA PVO charts |
| **Paper** | Derivative strips (`Jul26` / `Bal Month` / `Aug26` / balmo product, …) |

PVO quotes use **`C1_PRICE`** / **`C1_PRICE_BASIS`**.  
Reference HTMLs: `usld_cif_window/` (multiple dates).

```python
from datetime import date
from pathlib import Path

from build_ulsd_cif_report import build_ulsd_cif_report
from chart_data import discover_ulsd_cif_activity

REPORT_DATE = date(2026, 6, 25)  # try also 2026-06-05 / 2026-06-09 (busier)
OUTPUT_DIR = Path("output")
OUTPUT_DIR.mkdir(exist_ok=True)

print("Discovered instruments:")
display(discover_ulsd_cif_activity(REPORT_DATE))
```

```python
html = build_ulsd_cif_report(REPORT_DATE)
out = OUTPUT_DIR / f"ulsd_cif_window_{REPORT_DATE.isoformat()}.html"
out.write_text(html, encoding="utf-8")
print(f"Wrote {out} ({out.stat().st_size:,} bytes)")
```
